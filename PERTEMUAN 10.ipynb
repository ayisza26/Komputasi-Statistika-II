````python
scores = np.array([88, 72, 95, 80, 76])

print("Original:", scores)
print("Ascending:", np.sort(scores))
print("Descending:", np.sort(scores)[::-1])
print("Original remains:", scores)
````

````python
measurement_matrix = np.array([[8, 3, 6],
                               [9, 2, 5],
                               [4, 7, 1]])

print("Original matrix:")
print(measurement_matrix)
print("Sorted within each row:")
print(np.sort(measurement_matrix))
````

````python
print("Sorted by columns (axis=0):")
print(np.sort(measurement_matrix, axis=0))

print("Sorted by rows (axis=1):")
print(np.sort(measurement_matrix, axis=1))

print("All elements sorted:", np.sort(measurement_matrix, axis=None))
````

````python
students = np.array(["Ayu", "Budi", "Citra", "Doni"])
final_scores = np.array([81, 94, 76, 88])
order = np.argsort(final_scores)

print("Sorted scores:", final_scores[order])
print("Corresponding students:", students[order])
````

# Exercise 4 — Array Aggregation
Gunakan `visitors = np.array([[20, 25, 30], [18, 22, 28], [24, 27, 32]])`.

1. Hitung total dan rata-rata seluruh pengamatan.
2. Hitung total setiap kolom (`axis=0`) dan rata-rata setiap baris (`axis=1`).
3. Hitung jumlah kumulatif pada setiap baris (`np.cumsum`).
4. Hitung rata-rata setiap baris menggunakan `keepdims=True`, lalu periksa bentuknya.

````python
visitors = np.array([[20, 25, 30], [18, 22, 28], [24, 27, 32]])

# total dan rata2 seluruh pengamatan
print('Total:', np.sum(visitors))
print('Rata-rata:', np.mean(visitors), '\n')

# total setiap kolom dan rata2 setiapp baris
print('Total setiap kolom (axis = 0)    :', np.sum(visitors, axis = 0))
print('Rata-rata setiap baris (axis = 1):', np.mean(visitors, axis = 1), '\n')

# jumlah kumulatif setiap baris
print('Jumlah kumulatif setiap baris: \n', np.cumsum(visitors, axis = 1), '\n')

# hitung rata2 setiap baris
rowmean = np.mean(visitors, axis=1, keepdims=True)
print('Rata-rata setiap baris: \n', rowmean)
````

# Exercise 5 — Vectorized Operations
Gunakan `prices = np.array([10000, 15000, 20000, 25000])` dan `quantities = np.array([2, 3, 1, 4])`.

1. Hitung pendapatan tiap produk dengan perkalian element-wise.
2. Hitung total pendapatan menggunakan np.sum().
3. Berikan diskon 10% pada seluruh harga dengan operasi vektorisasi.
4. Tampilkan Boolean mask untuk produk dengan pendapatan minimal 50000.

````python
prices = np.array([10000, 15000, 20000, 25000])
quantities = np.array([2, 3, 1, 4])

total = prices * quantities
# pendapatan tiap produk
print("Pendapatan tiap produk (element-wise):", total)

# total pendapatan
print('Total pendapatan:', np.sum(total))

# diskon 10%
diskon = prices * 0.9
print('Harga setelah diskon 10%:', diskon)

# menampilkan boolean
mask = total >= 50000
print('Boolean untuk produk pendapatan min. 50000:', mask)
````

# Exercise 6 — Integrated Statistical Practice
Gunakan data nilai tiga ujian dari empat mahasiswa:

``` python
student_results = np.array([[80, 75, 90],
                            [65, 70, 72],
                            [92, 88, 95],
                            [78, 85, 82]])
```
1. Hitung rata-rata nilai masing-masing mahasiswa.
2. Pilih baris mahasiswa yang rata-ratanya minimal 80.
3. Tentukan posisi mahasiswa dengan nilai rata-rata tertinggi memakai np.argsort().
4. Tambahkan 5 poin pada setiap nilai, tetapi batasi nilai maksimum menjadi 100 dengan np.where().
5. Hitung rata-rata per ujian pada data yang telah diperbarui.

````python
student_results = np.array([[80, 75, 90],
                            [65, 70, 72],
                            [92, 88, 95],
                            [78, 85, 82]])

# rata2 nilai masing2 mahasiswa
rata2_nilai = np.round(np.mean(student_results, axis = 1), 2)
print('Rata-rata nilai masing-masing mhs:', rata2_nilai)

# baris mahasiswa yang rata2nya min 80
min80 = student_results[rata2_nilai >= 80]
print('Baris mhs yang rata-ratanya min. 80:\n', min80, '\n')

# posisi mahasiswa dengan nilai rata-rata tertinggi 
posisi = np.argsort(rata2_nilai)
print('Nilai rata-rata tertinggi:', posisi[-1])

# menambahkan 5 poin dengan batas maks. 100
poin5 = student_results + 5
hasil = np.where(poin5 > 100, 100, poin5)
print('Nilai dengan tambahan 5 poin:\n', hasil, '\n')

# rata2 per ujian
rataujian = np.mean(hasil, axis = 0)
print('Rata-rata per ujian (tambahan 5 poin):', rataujian)
````
